# Practice Question: Animal Inheritance in Python

## Problem Statement

Create a Python program to demonstrate **inheritance** using a parent class named `Animal` and two child classes named `Dog` and `Cat`.

The parent class should contain a common method that can be used by both child classes. Each child class should also have its own method to represent the sound made by that animal.

### Requirements

**1. Create the Parent Class: `Animal`**

- Create a class named `Animal`.
- Define a method named `eat()` inside the class.
- The `eat()` method should display a message indicating that the animal is eating.

**2. Create the Child Class: `Dog`**

- Create a class named `Dog` that inherits from the `Animal` class.
- Define a method named `sound()` inside the `Dog` class.
- The `sound()` method should display a message representing the sound made by a dog.

**3. Create the Child Class: `Cat`**

- Create a class named `Cat` that inherits from the `Animal` class.
- Define a method named `sound()` inside the `Cat` class.
- The `sound()` method should display a message representing the sound made by a cat.

**4. Create Objects and Call Methods**

- Create an object of the `Dog` class.
- Call the `eat()` method using the dog object.
- Call the `sound()` method using the dog object.
- Create an object of the `Cat` class.
- Call the `eat()` method using the cat object.
- Call the `sound()` method using the cat object.

### Expected Output

```text
Dog is eating.
Dog says: Woof! Woof!

Cat is eating.
Cat says: Meow! Meow!
```

### Your Task

Write the complete Python program using:

- A parent class named `Animal`.
- Two child classes named `Dog` and `Cat`.
- The `eat()` method in the parent class.
- The `sound()` method in each child class.
- Inheritance to reuse the `eat()` method.
- Objects to call the methods of both child classes.

### Important Note

Do not define the `eat()` method separately inside the `Dog` and `Cat` classes. Both child classes must inherit this method from the `Animal` parent class.

This exercise demonstrates **single inheritance and code reusability** in Python.

````python
class Atm:
    def __init__(self,balance):
        self.balance = balance

    def deposit(self,amount):
        print("First balance",self.balance)
        if amount>0:
            self.balance += amount
            print("Amount updated",self.balance)
        elif amount<0:
            print("We can not access amount")
    def withdrawl(self,amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Amount withdrwal",amount)
        elif amount > self.balance:
            print("Insufficient balance")
        else:
            print("Not correct integer")
    def check_balance(self):
        print("Current balance", self.balance)

a = Atm(5000)
a.deposit(2000)
a.withdrawl(5000)
a.withdrawl(5000)
a.check_balance()
````

````python
class Car:
    def __init__(self,brand,model,speed=0):
        self.brand = brand
        self.model = model
        self.speed = speed
    def accelerate(self,increment):
        if increment > 0:
            print("Initial speed", self.speed)
            self.speed += increment
            print("Current speed is",self.speed)
        elif increment<=0:
            print("Invalid speed")
    def display_details(self):
        print(self.brand)
        print(self.model)
        print(self.speed)
c = Car("tata","altroz")
c.accelerate(20)
c.accelerate(20)
c.display_details()
````

````python
class Animal:
    def eat(self):
        print("Animal is eating")
class Dog(Animal):
    def sound(self):
        print("The sound made by dog")
class Cat(Animal):
    def sound(self):
        print("The sound made by cat")
d = Dog()
d.eat()
d.sound()
c = Cat()
c.eat()
c.sound()
````
